# Typecasting in Python is the process of converting a value from one data type to another.

```python
x='10'
print(type(x))
```

```python
y="12"
print(type(y))
z=int(y)
print(type(z))
```

```python
number = '12'
print(type(number))
```

```python
# String -> Number
```

```python
new = int(number)
print(new)
print(type(new))
```

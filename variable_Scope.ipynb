# Scope means that the way which defines ways how we can access the variable.

#L ---->  E  ---->  G  ---->  B

```python
#Local Variable
def order():
    food = "Curd Rice"
    print("Your food is ", food)

order()
```

```python
#Enclosing

def Car():
    discount = 10
    def checkout():
        print("Applying discount : ", discount)
    checkout()

Car()
```

```python
# Enclosing means that a nested function can access variables defined in its outer function.
```

```python
#Global Scope

user_id = "anbu24"

def homepage():
    print("Welcome ", user_id)

def profile():
    print("Welcome to the profile page ", user_id)

homepage()
profile()
```
